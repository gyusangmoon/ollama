```python
from dotenv import load_dotenv
import os

from langchain_experimental.text_splitter import SemanticChunker
from langchain_huggingface import HuggingFaceEndpointEmbeddings

# .env 로드
load_dotenv()

# 환경변수 가져오기
hf_token = os.getenv("HUGGINGFACEHUB_API_TOKEN")

# HuggingFace Embeddings
embeddings = HuggingFaceEndpointEmbeddings(
    model="BAAI/bge-m3",
    huggingfacehub_api_token=hf_token,
)

# Semantic Chunker 연결
splitter = SemanticChunker(
    embeddings,
    breakpoint_threshold_type="percentile",
)

text = """
LangChain provides modular abstractions for LLM applications.

Semantic chunking splits text based on meaning similarity.

This is useful for RAG pipelines.

한국어 문서도 semantic chunking 으로 잘 나눌 수 있다.
"""

# 문서 생성
chunks = text_splitter.split_text(text)

# 출력
for i in enumerate(text):
    print(f"\n--- Chunk {i+1} ---")
    print(chunks[i].page_content)
```

```python
from langchain_experimental.text_splitter import SemanticChunker
from langchain_huggingface import HuggingFaceEndpointEmbeddings

# .env 로드
load_dotenv()

# 환경변수 가져오기
hf_token = os.getenv("HUGGINGFACEHUB_API_TOKEN")

# embedding (올바른 위치)
embeddings = HuggingFaceInferenceAPIEmbeddings(
    api_key=hf_token,
    model_name="BAAI/bge-m3"
)

# OpenAI 임베딩을 사용하여 의미론적 청크 분할기를 초기화합니다.
text_splitter = SemanticChunker(embeddings)

docs = text_splitter.create_documents(pdf_text)

print(len(docs))

for chunk in docs[120:180]:
    print(chunk.page_content)
    print('='*10)
```
